```python

import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq

# Load the API key
load_dotenv(r"C:\manojgenai\.env", override=True)

api_key = os.getenv("GROQ_API_KEY")

if not api_key:
    raise ValueError("Groq API key not found in .env")

# Initialize Groq
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    api_key=api_key,
    temperature=0
)

# Ask a question
response = llm.invoke("Explain Python in simple words.")

print(response.content)
```

```python
from langchain_core.messages import SystemMessage, HumanMessage

response = llm.invoke([
    SystemMessage(content="Think like you are mad and answer the question."),
    HumanMessage(content="Why is the sky blue?")
])

print(response.content)
```
