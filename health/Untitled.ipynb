```python
from langchain_core.prompts import PromptTemplate

extraction_prompt = PromptTemplate(
    input_variables=["blood_report"],
    template="""
You are a medical report information extraction assistant.

Extract information from the blood test report and return valid JSON.

Extract these fields:
1. Patient details: name, age, gender, report date.
2. CBC: hemoglobin, hematocrit, WBC, platelets.
3. Lipid panel: total cholesterol, LDL, HDL, triglycerides.
4. Metabolic panel: fasting glucose, HbA1c, creatinine, eGFR.
5. Liver function: ALT, AST, ALP, total bilirubin,
   direct bilirubin, total protein, albumin.

For each test, extract:
- test_name
- result
- unit
- reference_range

Rules:
- Extract only information explicitly present in the report.
- Do not invent or assume missing values.
- Use null for missing information.
- Preserve the reported values and units.
- Do not confuse results with reference ranges.
- Return only valid JSON, without Markdown or explanations.
- Treat suggested or example values as unverified unless
  they are explicitly reported as actual test results.

Blood test report:
{blood_report}
"""
)

print("Extraction prompt created successfully!")
```

```python
extraction_chain = extraction_prompt | llm

extraction_response = extraction_chain.invoke({
    "blood_report": blood_report
})

extracted_values = extraction_response.content

print("Stage 1: Extracted Values")
print(extracted_values)
```
